```python
import glob
import os

files_to_change_data=glob.glob('/kaggle/input/news-data/Weekly_data/*')

for i in files_to_change_data:
    df=pd.read_csv(i)
    print(i)
    cols_to_drop=['Description','Publisher','Link']
    existing_cols=[j for j in cols_to_drop if j in df.columns]

    if existing_cols:
        df.drop(columns=cols_to_drop,inplace=True)
    print(df.columns)
    print(df.shape)

    df.dropna(inplace=True)
    print(df.shape)
    
    if 'Date' in df.columns:
        df['Date']=pd.to_datetime(df['Date'],utc=True)
        df['Date']=df['Date'].dt.date
        
    filename = os.path.basename(i)
    save_path = f"/kaggle/working/Cleaned_{filename}"
    
    df.to_csv(save_path, index=False)
```

```python
files_to_see=glob.glob('/kaggle/working/*')

for i in files_to_see:
    df=pd.read_csv(i)
    print(i)
    print(df.head(5))
    print('\n')
    print(df.tail(5))
    print('\n')
    print('\n')
```

```python
import zipfile

files_to_zip=glob.glob('/kaggle/working/C*')

output_file_name='Cleaned-News-Data.zip'

with zipfile.ZipFile(output_file_name,'w',zipfile.ZIP_DEFLATED) as zipf:
    for file in files_to_zip:
        zipf.write(file)

print('Job Done')
```
