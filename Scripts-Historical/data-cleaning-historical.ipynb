```python
import glob
import os

files_to_change_data=glob.glob('/kaggle/input/historical-data/*')

for i in files_to_change_data:
    df=pd.read_csv(i)
    print(i)
    cols_to_drop=['Dividends','Stock Splits']
    existing_cols=[j for j in cols_to_drop if j in df.columns]

    if existing_cols:
        df.drop(columns=cols_to_drop,inplace=True)
    print(df.columns)
    print(df.shape)

    df.dropna(inplace=True)
    print(df.shape)
    
    if 'Date' in df.columns:
        df['Date']=pd.to_datetime(df['Date'],utc=True)
        df['Date']=df['Date'].dt.date
        
    filename = os.path.basename(i)
    save_path = f"/kaggle/working/Cleaned_{filename}"
    
    df.to_csv(save_path, index=False)
```
