```python
import yfinance as yf
import pandas as pd
import warnings

warnings.filterwarnings('ignore')
tickers = ['RELIANCE.NS', '^BSESN', '^NSEI', '^GSPC', '^IXIC', '^RUT']

for i in tickers:
    print(f"Downloading {i}...")
    data = yf.Ticker(i).history(period='max', auto_adjust=False)

    if data.empty:
        print(f"❌ No data found for {i}. Check your internet.")
    else:
        data = data[data['Volume'] > 0]
        clean_name = i.replace('^', '')
        filename = f"{clean_name}_Max.csv"
        
        data.to_csv(filename)
        
        print(f"✅ Data fetched successfully!")
        print(f"File saved as: {filename}")
        print(f"Shape: {data.shape}")
        print("-" * 30)
        
```
